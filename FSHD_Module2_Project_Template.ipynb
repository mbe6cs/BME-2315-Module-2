FSHD is a progressive muscular disease that can cause muscle weakness and degeneration over time. However, because disease progression varies between individuals and can occur gradually, measuring these changes accurately is challenging in clinical trials.

The goal of this project is to analyze FSHD patient data to compare MRI-derived muscle fat fraction and muscle strength measurements to find out which is more sensitive to disease progression. * Prevalence & incidence


FSHD is one of the most common muscular dystrophies in adults, affecting roughly 1 in 8,000 to 1 in 20,000 people. In the U.S., about 1 in 15,000 to 20,000 adults have it, and men and women are affected equally (Centro Dino Ferrari, n.d.; MedlinePlus, n.d.-a). A study in the Netherlands found about 52 new diagnoses per year and a prevalence of 12 per 100,000 people. The researchers believe the true number is higher, since many mild cases go undiagnosed. FSHD is inherited in an autosomal dominant pattern, so a child of an affected parent has a 50% chance of getting it, but 10–30% of cases appear with no family history (Deenen et al., 2014; Preston et al., 2020).

* Symptoms 
As the name suggests, FSHD mainly affects the face, shoulder blades, and upper arms. The weakness is often uneven between the two sides of the body and slowly spreads downward to the trunk and legs. Early signs include trouble closing the eyes, whistling, or making facial expressions. Symptoms usually start between ages 10 and 26, though some carriers never develop any, and 50–80% of patients experience chronic pain. The disease progresses slowly over decades. About 20% of patients eventually need a wheelchair, but life expectancy is usually normal (FSHD Society, n.d.; MedlinePlus, n.d.-a; Preston et al., 2020).

MedlinePlus. (n.d.). Facioscapulohumeral muscular dystrophy. U.S. National Library of Medicine. https://medlineplus.gov/ency/article/000707.htm

* Diagnosis
Doctors suspect FSHD from the pattern of weakness and family history, then confirm it with a genetic blood test. Both types involve a region on chromosome 4 called D4Z4, which normally keeps a gene called DUX4 switched off. In FSHD1 (about 95% of cases), this region is shortened. In FSHD2 (about 5%), it loses its normal “silencing,” usually because of a mutation in the SMCHD1 gene. In both cases, DUX4 becomes active and damages muscle cells. Muscle biopsies are now only used when genetic testing is inconclusive (Centro Dino Ferrari, n.d.; MedlinePlus, n.d.-b; Preston et al., 2020).

Standard of care treatments (& reimbursement) :
Currently, there are no cure or FDA approved disease modifying treatments for FSHD. Rather, standard care focuses on managing the symptoms, maintaining mobility, and improving the qualify of life. Physical therapy is highly recommended for those suffering with FSHD, with the intent of establishing appropriate exercise programs whilst assistive devices like ankle foot orthoses can improve mobility and reduce the risk of falls in patients experiencing foot drop. Additionally, occupational therapy may help patients experiencing difficulty with daily tasks. In specific cases, surgical fixation of the scapula can improve shoulder movement and chronic pain can be managed through physical therapy and medication. 

Reimbursement is more dependent on how much the users insurance covers and medical necessity. Under Originical Medicare, Part B covers medically necessary physical and occupational therapy. Upon meeting the Part B, deductible, patients will pay 20% of the Medicare-approved amount. Disease progression & prognosis :
FSHD is a slowly progressive muscular dystrophy, but the rate and the severity of muscle degeneration varies amongst individuals. Patients can experience periods of relatively stable muscle function followed by periods of rapid deterioration. Disease progression is asymmetric, muscles on the left and right side can differ in the rate at which they weaken. The age of onset can influence prognosis, with earlier onset cases associated with more severe disease progression. While FSHD progresses, healthy skeletal muscle tissue undergoes atrophy and can be replaced by fatty and fibrotic tissue, which reduces the muslces ability to generate force. 20% of affected individuals end up requiring wheelchair assistance, but typically, life expectancy is not shortened.
Because muscle degeneratoin occurs gradually and unevenly, it's harder to monitor disease progression. MRI provides a way to evaluate these changes in muscle composition, including fatty infiltration and signals connected with inflammation. The provided measurements help identify underlying changes that may not be immediately reflected in physical performance, making MRI valuable for monitoring disease propgession and for evaluating what treatments are needed for clinical trials.

Biological mechanisms (anatomy, organ physiology, cell & molecular physiology) :
FSHD affects skeletal muscles, those responsible for voluntary movement and generating force. Skeletal muslces consist of bundles of muscle fibers that have myofibrils, which are made up of sarcomeres. These sarcomeres have actin and myosin filaments interacting to produce muscle contractions. When FSHD damages muslce fibers, the amount of functional contractile tissue decreases, which reduces the muscle's ability to generate force and perform normal acts.
FSHD is caused by the abnormal expression of the DUX4 gene, which is active during the early embryonic development, but is silenced within the mature skeletal muscle. There are two primary forms of FSHD: FSHD1, which makes up 95% of cases, results from a contraction of the D4Z4 DNA repeat region on chromosome 4, leading to reduced DNA methylation and inappropriate DUX4 expression. FSHD2, which makes up the remaining 5% of cases, is associated with mutations in the SMCHD1 gene, which disrupts the normal epigenetic silencing of DUX4. 
DUX4 encodes a transcription factor that influences the expression of other genes. When its abnormally expressed within skeletal muscle, it will activate gene programs that contribute to cellular dysfunction, inflammation, and muscle damage. Over time, this leads to muscle atrophy and the replacing of functional muscle fibers. But, these changes can be observed through MRI, which provides a connection between the molecular cause of FSHD and the changes in muscle composition.

Clinical Trials/next-gen therapies :
The current clinical trials for FSHD involve research into therapies targeting the molecular basis of muscle degradation, mainly the deregulation of DUX4 expression. An example of a therapy that was studied in the past is Losmapimod, an experimental drug from Fulcrum Therapeutics that was made to regulate DUX4 expression through inhibition of p38 MAPK pathway. While it was promising in earlier studies, it didn't prove to show a statistically significant improvement in its primary outcome, the Reachable Workspace in its Phase 3 REACH clinical trial and then was discontinued in 2024. Other drugs under investigation include del-brax AOC 1020, an experimental RNA-targeted therapeutic meant to lower DUX4 expression, and EPI-321, an epigenetic editing therapeutic aimed at restoring DUX4 silencing. Though they are still under clinical investigation, both have shown signs of biological activity without proving their clinical efficacy. This highlights the need for selection of sensitive and appropriate endpoints during clinical trials.
