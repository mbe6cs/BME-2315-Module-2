## Project Goal:
FSHD is a progressive muscular disease that can cause muscle weakness and degeneration over time. However, because disease progression varies between individuals and can occur gradually, measuring these changes accurately is challenging in clinical trials.

The goal of this project is to analyze FSHD patient data to compare MRI-derived muscle fat fraction and muscle strength measurements to find out which is more sensitive to disease progression. By examining how these measurements change over time, we want to identify which gives us a more consistent and reliable indication of muscle degeneration. Our findings will help us recommend which measurement may be more useful when evaluating new FSHD treatments in clinical trials.

Research Question: Which is more sensitive to FSHD disease progression, MRI fat fraction or muscle strength?

## Disease Background:

*The following 7 bullets seem most relevant, but feel free to choose others if you like:*

* Prevalence & incidence

FSHD is one of the most common muscular dystrophies in adults, affecting roughly 1 in 8,000 to 1 in 20,000 people. In the U.S., about 1 in 15,000 to 20,000 adults have it, and men and women are affected equally (Centro Dino Ferrari, n.d.; MedlinePlus, n.d.-a). A study in the Netherlands found about 52 new diagnoses per year and a prevalence of 12 per 100,000 people. The researchers believe the true number is higher, since many mild cases go undiagnosed. FSHD is inherited in an autosomal dominant pattern, so a child of an affected parent has a 50% chance of getting it, but 10–30% of cases appear with no family history (Deenen et al., 2014; Preston et al., 2020).

* Symptoms 

As the name suggests, FSHD mainly affects the face, shoulder blades, and upper arms. The weakness is often uneven between the two sides of the body and slowly spreads downward to the trunk and legs. Early signs include trouble closing the eyes, whistling, or making facial expressions. Symptoms usually start between ages 10 and 26, though some carriers never develop any, and 50–80% of patients experience chronic pain. The disease progresses slowly over decades. About 20% of patients eventually need a wheelchair, but life expectancy is usually normal (FSHD Society, n.d.; MedlinePlus, n.d.-a; Preston et al., 2020).


* Diagnosis

Doctors suspect FSHD from the pattern of weakness and family history, then confirm it with a genetic blood test. Both types involve a region on chromosome 4 called D4Z4, which normally keeps a gene called DUX4 switched off. In FSHD1 (about 95% of cases), this region is shortened. In FSHD2 (about 5%), it loses its normal “silencing,” usually because of a mutation in the SMCHD1 gene. In both cases, DUX4 becomes active and damages muscle cells. Muscle biopsies are now only used when genetic testing is inconclusive (Centro Dino Ferrari, n.d.; MedlinePlus, n.d.-b; Preston et al., 2020).


* Standard of care treatments (& reimbursement) :

Currently, there are no cures or FDA-approved disease-modifying treatments for FSHD. Rather,
standard care focuses on managing the symptoms, maintaining mobility, and improving the
quality of life (Preston & Wang, 2025). Physical therapy is highly recommended for those
suffering from FSHD, with the intent of establishing appropriate exercise programs, whilst
assistive devices like ankle-foot orthoses can improve mobility and reduce the risk of falls in
patients experiencing foot drop. Additionally, occupational therapy may help patients
experiencing difficulty with daily tasks. In specific cases, surgical fixation of the scapula can
improve shoulder movement, and chronic pain can be managed through physical therapy and
medication (Preston & Wang, 2025).
Reimbursement is more dependent on how much the user's insurance covers and medical
necessity. Under Original Medicare, Part B covers medically necessary physical and
occupational therapy. Upon meeting the Part B deductible, patients will pay 20% of the
Medicare-approved amount (Centers for Medicare & Medicaid Services, 2026). Coverage and
out-of-pocket costs for surgical procedures and assistive devices depend on their specific
coverage requirement.

* Disease progression & prognosis : 

FSHD is a slowly progressive muscular dystrophy, but the rate and the severity of muscle
degeneration vary amongst individuals. Patients can experience periods of relatively stable
muscle function followed by periods of rapid deterioration. Disease progression is asymmetric;
muscles on the left and right sides can differ in the rate at which they weaken (Preston & Wang,
2025). The age of onset can influence prognosis, with earlier onset cases associated with more
severe disease progression. While FSHD progresses, healthy skeletal muscle tissue undergoes
atrophy and can be replaced by fatty and fibrotic tissue, which reduces the muscles' ability to
generate force. 20% of affected individuals end up requiring wheelchair assistance, but typically,
life expectancy is not shortened (Preston & Wang, 2025). Because muscle degeneration occurs
gradually and unevenly, it's harder to monitor disease progression. MRI provides a way to
evaluate these changes in muscle composition, including fatty infiltration and signals connected
with inflammation. The provided measurements help identify underlying changes that may not
be immediately reflected in physical performance, making MRI valuable for monitoring disease
progression and for evaluating what treatments are needed for clinical trials (Wong et al., 2024).

* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology) :

FSHD affects skeletal muscles, those responsible for voluntary movement and generating force.
Skeletal muscles consist of bundles of muscle fibers that have myofibrils, which are made up of
sarcomeres. These sarcomeres have actin and myosin filaments interacting to produce muscle
contractions. When FSHD damages muscle fibers, the amount of functional contractile tissue
decreases, which reduces the muscle's ability to generate force and perform normal actions
(Preston & Wang, 2025). FSHD is caused by the abnormal expression of the DUX4 gene, which
is active during early embryonic development, but is silenced within the mature skeletal muscle.
There are two primary forms of FSHD: FSHD1, which makes up 95% of cases, results from a
contraction of the D4Z4 DNA repeat region on chromosome 4, leading to reduced DNA
methylation and inappropriate DUX4 expression. FSHD2, which makes up the remaining 5% of
cases, is associated with mutations in the SMCHD1 gene, which disrupts the normal epigenetic
silencing of DUX4 (MedlinePlus Genetics, n.d.; Preston & Wang, 2025). DUX4 encodes a
transcription factor that influences the expression of other genes. When it is abnormally
expressed within skeletal muscle, it will activate gene programs that contribute to cellular
dysfunction, inflammation, and muscle damage. Over time, this leads to muscle atrophy and the
replacement of functional muscle fibers. But these changes can be observed through MRI,
which provides a connection between the molecular cause of FSHD and the changes in muscle
composition (Wong et al., 2024).

* Clinical Trials/next-gen therapies :

The current clinical trials for FSHD involve research into therapies targeting the molecular basis
of muscle degradation, mainly the deregulation of DUX4 expression (Wong et al., 2024). An
example of a therapy that was studied in the past is Losmapimod, an experimental drug from
Fulcrum Therapeutics that was made to regulate DUX4 expression through inhibition of the p38
MAPK pathway. While it was promising in earlier studies, it didn't prove to show a statistically
significant improvement in its primary outcome, the Reachable Workspace in its Phase 3
REACH clinical trial and then was discontinued in 2024 (Kinoshita, 2024; Voermans et al.,
2026). Other drugs under investigation include del-brax AOC 1020, an experimental
RNA-targeted therapeutic meant to lower DUX4 expression, and EPI-321, an epigenetic editing
therapeutic aimed at restoring DUX4 silencing (Novartis, 2026; UCLA Health, 2026). Though
they are still under clinical investigation, both have shown signs of biological activity without
proving their clinical efficacy (Novartis, 2026; Epicrispr Biotechnologies, 2026).This highlights
the need for selection of sensitive and appropriate endpoints during clinical trials.

Describe the FSHD dataset used in your project. Include:

For this project, we used the provided FSHD dataset, which includes information from 34 participants with measurements collected at baseline, Year 1, and Year 2. The clinical dataset has patient information such as age, sex, and D4Z4 repeat length, along with MRI measurements, muscle strength, and functional assessments. - The subject and visit/timepoint(s) analyzed

We performed two analyses. For the clinical analysis, we used the 21 of 34 participants who had complete right-leg TA fat fraction and dorsiflexion strength at baseline, Year 1, and Year 2. Because visit times were different  (Year 1: 11.15–18.18 months; Year 2: 22.13–42.77 months), we used linear interpolation to estimate each participant’s values at 11.15 and 22.13 months, the earliest visit times. That way every estimate falls within the measured data. For the MRI image analysis, we examined eight slices of the left TA from subject UW_0020 at two visits, TP1 and TP2. This subject   matches participant 1-0020 in the clinical dataset, whose Year 1 visit was 16.8 months after baseline. This participant had no Year 2 data, so they were not part of the 21-participant comparison. The MRI analysis was used to show how fat fraction is calculated and how it varies along the muscle and between visits.

- The tibialis anterior MRI data

The tibialis anterior is the muscle at the front of the lower leg that lifts the foot (dorsiflexion). Weakness in this muscle causes foot drop, which is common in FSHD. Wong et al. (2024) imaged and biopsied the TA on both legs of every participant. That makes it the one muscle where MRI, strength, and molecular data are all available.

- The water-image and fat-image files

The MRI teaching dataset contains water-signal and fat-signal images saved as NumPy .npy files. These come from a Dixon MRI scan, which separates the signal coming from water which is mostly muscle tissue from the signal coming from fat. Each image is 266 × 224 pixels and shows the half of the leg cross-section that contains the TA, along with some surrounding anatomy and background. According to dataset_info.json, each pixel is 1.116 × 1.116 mm and each slice is 4 mm thick.

- The tibialis anterior segmentation mask

 Each slice has a binary mask in which 1 = TA pixel and 0 = everything else. The masks were provided with the dataset; we did not create or edit them. We used the mask to keep only TA pixels when calculating fat fraction, so that background, bone, subcutaneous fat, and other muscles would not affect the result.

- The sampled locations along the muscle

The MRI dataset had eight sampled locations along the tibialis anterior muscle at each timepoint, corresponding to 10%, 20%, 30%, 40%, 50%, 60%, 70%, and 80% of the way along the muscle. The provided slice_info.csv file identifies each slice's timepoint, percentage location, and distance from the start of the muscle in millimeters. The same percentage location did not always correspond to the same physical distance at both visits. For example, the 40% slice was located at 92 mm at TP1 and 96 mm at TP2. Therefore, we used distance in millimeters rather than percentage location for our interpolation analysis. We calculated fat fraction at the available locations for TP1 and TP2 and used linear interpolation to estimate fat fraction at 100 mm along the muscle. This allowed us to compare fat fraction at the same distance from the muscle start between the two timepoints.

- The meaning and units of the image signals

The water and fat images contain MRI signal intensities in the scanner’s own arbitrary units, so the raw numbers don’t mean much by themselves. We combined them as fat fraction (%) = fat / (fat + water) × 100. A value near 0% means mostly healthy muscle and a value near 100% means the muscle has mostly been replaced by fat. Outside the leg there is almost no signal, so the ratio there is just noise, which is why we only used pixels inside the TA mask.

- The clinical, biomarker, or functional measurements included in the dataset

The clinical dataset has 34 participants who were measured at baseline, Year 1, and Year 2, and it records how many months after baseline each visit happened. For each person it lists sex, age, and D4Z4 repeat size. On the MRI side, it gives the whole-muscle fat fraction (%) of the tibialis anterior (TA) on both legs, plus a STIR value that shows inflammation or swelling in the muscle which is only at baseline and Year 1. There is also a DUX4 score from the TA biopsies taken at baseline, which shows how active the genes turned on by DUX4 are. Strength was measured as how hard each foot could pull upward (dorsiflexion), in kilograms. The dataset also has timed tests of function in seconds: the Timed Up-and-Go, a 10-meter walk/run, and going up and down four stairs. Finally, the Clinical Severity Score rates the disease from 0 to 10, with higher scores meaning more severe. We chose TA fat fraction and TA strength for our analysis because they measure the same muscle. We also checked that their ranges in the dataset (3.3–82.2% and 1.8–43 kg) match the ones in Wong et al. (2024).

- The source of the data

Both datasets were provided through BME 2315 Module 2 and come from the study by Wong et al. (2024). That study enrolled 34 FSHD participants (16 female, 18 male, ages 21–69) at the University of Washington, University of Rochester, and University of Kansas. Participants had lower-leg MRI on 3T Siemens scanners using a Dixon sequence (1.1 × 1.1 × 4 mm voxels), strength testing of the TA in kilograms, and biopsies of both TA muscles. The TA was outlined on every MRI slice by an AI program and then checked by trained staff, and whole-muscle fat fraction was calculated as the average pixel fat fraction across the whole TA. According to the README, the MRI teaching files are cross-sections of the left TA of subject UW_0020 at two visits, each with a water image, a fat image, and a mask. The masks were provided, so we did not create them ourselves.

We used Python to analyze the MRI data provided for this project. We used the NumPy library to load the water-signal images, fat-signal images, and segmentation masks, which were saved as .npy files. We also used Pandas to read the slice_info.csv file, which has information about where each MRI slice was taken along the muscle. We are focusing on the left tibialis anterior (TA) muscle of subject UW_0020 at two visits, TP1 and TP2, with eight sampled MRI slices from each visit.

- How the tibialis anterior mask was used to identify the muscle

We are using the provided segmentation mask to identify the TA muscle in each MRI image. The mask has values of 1 for pixels inside the TA muscle and 0 for pixels outside the muscle. By selecting only the pixels labeled 1, we were able to separate the TA muscle from surrounding tissues. This is important because including surrounding fat or other muscles could affect our fat-fraction measurements.

- How fat fraction was calculated for the whole image region and for the TA muscle only
We used the following formula to calculate how much fat is in the MRI images using the fat and water signals.

Fat Fraction (%) = [Fat Signal / (Fat Signal + Water Signal)] × 100

First, we  calculated the fat percentage in the whole MRI image. Then, we used the segmentation mask to calculate the fat percentage only inside the TA muscle. We also compared the results to see how surrounding tissues affect the measurements.

- How the fat-fraction distribution was summarized

We summarized the distribution of fat fraction within the TA in two ways. - Which summary statistics were calculated, such as mean, median, standard deviation, and number of muscle pixels
For the TA in TP1, slice 01, we calculated the mean (83.92%), median (88.83%), standard deviation (13.27), and number of muscle pixels (61). The mean gives the average fat fraction, the median gives the middle value and is less affected by extreme pixels, the standard deviation shows how much fat fraction varies within the muscle, and the pixel count shows how large the TA region is in that slice. - How measurements were summarized across locations along the muscle
For each of the eight slices at each visit, we calculated the mean fat fraction of TA pixels and plotted it against the slice’s location along the muscle. If you compare visits or use interpolation in your final analysis, describe those methods clearly, including the variables used and the assumptions made.

We used linear interpolation in two ways. For the MRI data, we interpolated across distance: because the same percentage location did not always match the same physical distance at both visits (for example, the 40% slice was at 92 mm at TP1 and 96 mm at TP2), we used distance_from_muscle_start_mm and np.interp() to estimate fat fraction at 100 mm at both visits (visit_label = TP1 or TP2; target = 100 mm). For the clinical data, we interpolated across time: because visit times varied, we used scipy’s interp1d to estimate each participant’s right TA fat fraction (ff_R) and dorsiflexion strength (dorsiflexion_qmt_R) at common time points T1 = 11.15 months and T2 = 22.13 months. Only the 21 participants with all three visits for both measures were included. We then compared percent change from T1 to T2 using the mean, median, standard deviation, signal-to-noise ratio (mean ÷ SD), and a one-sample t-test. Both methods assume values change at a constant rate between measurements. We tested this by repeating the clinical analysis with quadratic interpolation.
```python
#importing the libraries we need for this analysis
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

#Find the folder containing the MRI data
folder = Path("FSHD_MRI_data") / "FSHD_TA_MRI_data_0020_TP1_TP2"
#load all the images and masks for the first MRI slice
water = np.load(folder / "TP1_slice_01_water.npy").astype(float)
fat = np.load(folder / "TP1_slice_01_fat.npy").astype(float)
mask = np.load(folder / "TP1_slice_01_mask.npy") > 0

#Identify the TA muscle and calculate fat fraction
total = water + fat
valid = np.isfinite(total) & (total > 0)

ff = np.full(total.shape, np.nan)
ff[valid] = fat[valid] / total[valid] * 100

#separate the fat fraction values for the whole image and the TA muscle
whole = ff[valid]
muscle = ff[mask & valid]

#Calculate summary statistics
print("Whole image mean:", round(np.mean(whole), 2), "%")
print("TA mean:", round(np.mean(muscle), 2), "%")
print("TA median:", round(np.median(muscle), 2), "%")
print("TA standard deviation:", round(np.std(muscle), 2))
print("TA muscle pixels:", np.count_nonzero(mask))

#Display the MRI fat fraction map
plt.figure(figsize=(6, 5))
plt.imshow(np.where(mask, ff, np.nan),
           cmap="viridis", vmin=0, vmax=100)
plt.colorbar(label="Fat Fraction (%)")
plt.title("TA Fat Fraction - TP1, Slice 01")
plt.xlabel("Pixel Column")
plt.ylabel("Pixel Row")
plt.show()

#display fat fraction distribution
plt.figure(figsize=(6, 4))
plt.hist(muscle, bins=15, edgecolor="black")
plt.title("TA Fat Fraction Distribution")
plt.xlabel("Fat Fraction (%)")
plt.ylabel("Number of Pixels")
plt.show()

#Compare MRI slices along the muscle
slices = []
means = []

for file in sorted(folder.glob("TP1_slice_*_water.npy")):
    prefix = file.stem.replace("_water", "")

    w = np.load(file).astype(float)
    f = np.load(folder / f"{prefix}_fat.npy").astype(float)
    m = np.load(folder / f"{prefix}_mask.npy") > 0
#only include pixels that have valid measurements
    total = w + f
    valid_pixels = m & np.isfinite(total) & (total > 0)
#if the slice doesnt have valid pixels you can skip it
    if not np.any(valid_pixels):
        continue

    values = f[valid_pixels] / total[valid_pixels] * 100
    slices.append(int(prefix.split("_")[2]))
    means.append(np.mean(values))
#plot the mean fat fraction along the TA muscle
plt.figure(figsize=(7, 4))
plt.plot(slices, means, marker="o")
plt.title("Mean Fat Fraction Along the TA Muscle")
plt.xlabel("MRI Slice Number")
plt.ylabel("Mean Fat Fraction (%)")
plt.grid(True)
plt.show()


```

```python
import pandas as pd

#Load information about all MRI slices
info = pd.read_csv(folder / "slice_info.csv")

plt.figure(figsize=(8, 5))

# TP1 and TP2 represent separate MRI visits
#analyze both MRI visits and plot the mean fat fraction along the TA muscle
for visit_label in ["TP1", "TP2"]:

    locations = []
    means = []

    visit = info[info["timepoint"] == visit_label]

    for _, row in visit.iterrows():

        water = np.load(folder / row["water_file"]).astype(float)
        fat = np.load(folder / row["fat_file"]).astype(float)
        mask = np.load(folder / row["mask_file"]) > 0

        total = water + fat
        valid = mask & np.isfinite(total) & (total > 0)

        ff = fat[valid] / total[valid] * 100

        locations.append(row["percent_along_muscle"])
        means.append(np.mean(ff))

    plt.plot(locations, means, marker="o", label=visit_label)
#plot the mean fat fraction along the TA muscle for both MRI visits
plt.title("TA Fat Fraction: TP1 vs TP2")
plt.xlabel("Location Along Muscle (%)")
plt.ylabel("Mean Fat Fraction (%)")
plt.legend()
plt.grid(True)
plt.show()
```

## Interpolation Analysis:

For this part of our Interpolation Analysis, we used linear interpolation to estimate the fat fraction at a location along the tibialis anterior (TA) muslce where we did not have a direct sampled MRI slice. From the data set, we used the distance_from_muscle_start_mm column from slice_info.csv to identify each of the slice's location in mm. Specifically, for subject UW_0020, we analysed eight of the sampled location at both TP1 and TP2, and calculated the average fat fraction within the TA muslce, respective to it's location.

We chose to estimate the fat fraction at 100mm along the muscle using np.interp(). Since 100mm fell between two measured location at both visits, we performed interpolation rather than extrapolation. At TP1, the surrounding measured locations were 92mm and 116mm, and at TP2, they were 96mm and 120mm. Linear interpolation assumes that the fat fraction changes at a constant rate between the neighboring measurements.

Through this method, we estimated a fat fraction of 26.91% at TP1 and 52.40% at TP2 at 100mm. Our graph shows the measured mean fat fractions along the muscle, with diamond shaped markers representing the interpolated estimates. The results suggest that fat fraction increased between the two visits at this location, indicating increased fat replacement within the muscle.

But, these interpolated values are estimates rather than new MRI measurements. The actual fat fraction at 100mm can differ because fat may not have been distributed evenly between the sampled locations. Additionally, differences in image quality, the muscle segmentation, and the limited number of sampled slices can affect the accuracy of our estimates.

```python
#Linear interpolation at 100 mm
target = 100

plt.figure(figsize=(8, 5))

# TP1 and TP2 represent separate MRI visits
for visit_label in ["TP1", "TP2"]:

    visit = info[info["timepoint"] == visit_label]
    locations = []
    means = []

    for _, row in visit.iterrows():
        water = np.load(folder / row["water_file"]).astype(float)
        fat = np.load(folder / row["fat_file"]).astype(float)
        mask = np.load(folder / row["mask_file"]) > 0

        total = water + fat
        valid = mask & np.isfinite(total) & (total > 0)

        ff = fat[valid] / total[valid] * 100

        locations.append(row["distance_from_muscle_start_mm"])
        means.append(np.mean(ff))

    #sort the locations and means for interpolation by distance
    locations, means = zip(*sorted(zip(locations, means)))

    #Interpolate
    estimate = np.interp(target, locations, means)

    print(f"{visit_label} at 100 mm: {estimate:.2f}%")

    plt.plot(locations, means, marker="o", label=visit_label)
    plt.scatter(target, estimate, marker="D", s=100)
plt.xlabel("Distance Along Muscle (mm)")
plt.ylabel("Mean Fat Fraction (%)")
plt.title("Linear Interpolation of TA Fat Fraction")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

#Load the patient dataset
df = pd.read_csv("Module2_FSHD_Dataset.csv")

#Select common time points
T1 = df["lag_months_y1"].min()
T2 = df["lag_months_y2"].min()

print("Number of patients:", len(df))
print("Year 1 common time:", T1)
print("Year 2 common time:", T2)

#Display the measurements we're comparing
print(df[[
    "subject_id",
    "ff_baseline_R",
    "ff_y1_R",
    "ff_y2_R",
    "dorsiflexion_qmt_R_baseline",
    "dorsiflexion_qmt_R_y1",
    "dorsiflexion_qmt_R_y2"
]].head())
```

```python
#Select the patients with complete MRI and QMT measurements
time_cols = ["lag_months_baseline", "lag_months_y1", "lag_months_y2"]

ff_cols = ["ff_baseline_R", "ff_y1_R", "ff_y2_R"]

qmt_cols = [
    "dorsiflexion_qmt_R_baseline",
    "dorsiflexion_qmt_R_y1",
    "dorsiflexion_qmt_R_y2"
]

complete = df.dropna(subset=time_cols + ff_cols + qmt_cols).copy()

#Function for linear interpolation
def interpolate_patient(row, columns, target):
    times = [row[col] for col in time_cols]
    values = [row[col] for col in columns]

    f = interp1d(times, values, kind="linear")
    return float(f(target))

#Interpolate both measurements at T1 and T2
for name, columns in [("FF", ff_cols), ("QMT", qmt_cols)]:
    complete[name + "_T1"] = complete.apply(
        lambda row: interpolate_patient(row, columns, T1), axis=1
    )
    complete[name + "_T2"] = complete.apply(
        lambda row: interpolate_patient(row, columns, T2), axis=1
    )
#print the number of patients included in the analysis and display the first few rows of the relevant data
print("Patients included:", len(complete))
#display the first few rows of the data with the interpolated values for FF and QMT at T1 and T2
display(complete[[
    "subject_id", "FF_T1", "FF_T2", "QMT_T1", "QMT_T2"
]].head())
```

```python
#Calculate the percentage change from T1 to T2
complete["FF_change"] = (
    (complete["FF_T2"] - complete["FF_T1"])
    / complete["FF_T1"]
) * 100

complete["QMT_change"] = (
    (complete["QMT_T2"] - complete["QMT_T1"])
    / complete["QMT_T1"]
) * 100

#Display each of the patient's changes
display(complete[[
    "subject_id", "FF_change", "QMT_change"
]].round(2))

#Calculate mean and standard deviation
print("MRI Fat Fraction:")
print("Mean % change:", round(complete["FF_change"].mean(), 2))
print("Standard deviation:", round(complete["FF_change"].std(), 2))

print("\nMuscle Strength (QMT):")
print("Mean % change:", round(complete["QMT_change"].mean(), 2))
print("Standard deviation:", round(complete["QMT_change"].std(), 2))
```

```python
#Compare MRI fat fraction and muscle strength

ff_mean = complete["FF_change"].mean()
qmt_mean = complete["QMT_change"].mean()

ff_sd = complete["FF_change"].std()
qmt_sd = complete["QMT_change"].std()

#Create the comparison graph
plt.figure(figsize=(8, 5))

plt.bar(
    ["MRI Fat Fraction", "Muscle Strength (QMT)"],
    [ff_mean, qmt_mean],
    yerr=[ff_sd, qmt_sd],
    capsize=6
)

plt.ylabel("Mean Change (%)")
plt.title("FSHD Progression: MRI vs Muscle Strength (n=21)")
plt.axhline(0, color="black", linewidth=1)
plt.tight_layout()
plt.show()

#Compare change relative to variability of each measurement
print("MRI Mean / SD:", round(ff_mean / ff_sd, 2))
print("QMT Mean / SD:", round(qmt_mean / qmt_sd, 2))
```

```python
#Visualize linear vs. quadratic interpolation for only one patient
patient = complete.iloc[0]

times = [patient[col] for col in time_cols]
values = [patient[col] for col in ff_cols]

#Create interpolation functions
linear = interp1d(times, values, kind="linear")
quadratic = interp1d(times, values, kind="quadratic")

#Create plotting points
months = np.linspace(times[0], times[2], 100)

plt.figure(figsize=(8, 5))

plt.scatter(times, values, color="black",
            label="Actual MRI measurements", zorder=3)

plt.plot(months, linear(months),
         label="Linear interpolation")

plt.plot(months, quadratic(months),
         linestyle="--", label="Quadratic interpolation")

#Show estimates at T1 and T2
plt.scatter([T1, T2], [linear(T1), linear(T2)],
            marker="D", s=80, label="Linear estimates")

plt.xlabel("Time (months)")
plt.ylabel("MRI Fat Fraction (%)")
plt.title("Linear vs Quadratic Interpolation: " + patient["subject_id"])
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

print("Linear estimate at T2:", round(float(linear(T2)), 2))
print("Quadratic estimate at T2:", round(float(quadratic(T2)), 2))
```

```python
from scipy import stats

#compute T-test for MRI fat fraction change
ff_t, ff_p = stats.ttest_1samp(
    complete["FF_change"], 0
)

# compute T-test for muscle strength change
qmt_t, qmt_p = stats.ttest_1samp(
    complete["QMT_change"], 0
)

print("MRI Fat Fraction:")
print("T-statistic:", round(ff_t, 3))
print("P-value:", round(ff_p, 4))

print("\nMuscle Strength (QMT):")
print("T-statistic:", round(qmt_t, 3))
print("P-value:", round(qmt_p, 4))
```

```python
#This is to compare linear and quadratic interpolation for all 21 patients

def quadratic_change(row, columns):
    times = [row[col] for col in time_cols]
    values = [row[col] for col in columns]

    f = interp1d(times, values, kind="quadratic")

    value_T1 = float(f(T1))
    value_T2 = float(f(T2))

    return (value_T2 - value_T1) / value_T1 * 100

#Calculate quadratic percentage changes
complete["FF_quad_change"] = complete.apply(
    lambda row: quadratic_change(row, ff_cols), axis=1
)

complete["QMT_quad_change"] = complete.apply(
    lambda row: quadratic_change(row, qmt_cols), axis=1
)

#Compare average changes
print("MRI Fat Fraction:")
print("Linear mean:", round(complete["FF_change"].mean(), 2))
print("Quadratic mean:", round(complete["FF_quad_change"].mean(), 2))

print("\nMuscle Strength (QMT):")
print("Linear mean:", round(complete["QMT_change"].mean(), 2))
print("Quadratic mean:", round(complete["QMT_quad_change"].mean(), 2))

#Visualize the comparison
plt.figure(figsize=(8, 5))

plt.scatter(
    complete["FF_change"],
    complete["FF_quad_change"],
    label="MRI Fat Fraction"
)

plt.scatter(
    complete["QMT_change"],
    complete["QMT_quad_change"],
    label="Muscle Strength (QMT)"
)

plt.xlabel("Linear Interpolation: % Change")
plt.ylabel("Quadratic Interpolation: % Change")
plt.title("Linear vs Quadratic Interpolation (n=21)")
plt.axline((0, 0), slope=1, linestyle="--", color="gray")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## Conclusions:

Our research question was “Which is more sensitive to FSHD disease progression, MRI fat fraction or muscle strength?” Our analysis indicates that MRI fat fraction is more sensitive to disease-related change than muscle strength. Fat fraction increased in most participants, while muscle strength was highly variable and showed no statistically significant change.

Which measures changed most consistently: Among the 21 participants with complete data, tibialis anterior fat fraction increased by an average of 5.88% (SD = 12.38%). It increased in 15 of 21 participants, and the change was statistically significant (p = 0.0417). Muscle strength had a mean change of +4.98% (SD = 20.24%), but it rose in 11 participants and fell in 10, and the change was not statistically significant (p = 0.2732). This does not mean FSHD patients were getting stronger. FSHD is a progressive disease that causes muscle weakness, and the positive average was driven by two participants whose strength changed by +54.10% and +49.20%. Without these two participants, the remaining 19 had an average strength change of about 0.07%, essentially no change. Strength testing depends on patient effort and positioning, so it varies from visit to visit by more than FSHD’s slow decline over two years. As a result, strength testing could not detect disease progression during this period, while MRI fat fraction could.

Which measures were most sensitive to differences across visits and locations: To compare sensitivity, we calculated the signal-to-noise ratio by dividing the mean percent change by its standard deviation. Fat fraction’s ratio was nearly twice that of muscle strength (0.47 vs. 0.25), meaning its change was larger relative to the variation between participants. A higher signal-to-noise ratio means real change is easier to separate from measurement noise. In a clinical trial, this means fewer participants would be needed to detect a treatment effect, making MRI fat fraction a more efficient way to test whether new FSHD therapies work. MRI also detected differences along the muscle, which strength testing cannot measure. In subject UW_0020, the estimated fat fraction at 100 mm along the tibialis anterior increased from 26.91% at TP1 to 52.40% at TP2, an increase of 25.49 percentage points. This shows that fat replacement can progress quickly and unevenly within a single muscle. Because this analysis included only one subject, it shows what MRI can detect rather than a pattern for all patients.

Usefulness as biomarkers or clinical-trial endpoints: MRI fat fraction is a strong candidate for a biomarker or exploratory endpoint in FSHD clinical trials. It provides an objective, quantitative measure of muscle degeneration, and it can detect progression when strength measurements remain stable or fluctuate between visits. Muscle strength testing remains important because it reflects how patients function in daily life. Using both together gives a more complete picture of disease progression.

Additional evidence needed: When we repeated the analysis using quadratic interpolation, the average changes remained similar (5.42% for fat fraction and 3.05% for strength), so our findings were not driven by the choice of interpolation method. However, our study included only 21 participants, about two years of follow-up, and one muscle. Before either measure is used to evaluate treatment effectiveness, larger studies with longer follow-up and more muscles are needed to confirm these results. Researchers would also need to show that changes in MRI fat fraction correspond to meaningful changes in strength and daily function, and that fat fraction responds to effective treatment.

Ethical implications: Choosing between MRI fat fraction and strength testing involves cost, accessibility, and patient relevance. MRI is more sensitive but expensive and less accessible, which could exclude some patients from clinical trials. Strength testing is cheaper and directly tied to daily life, but because it is less sensitive, it could miss real progression and make an effective drug appear to fail. Relying only on MRI carries the opposite risk: a treatment could improve the scan without improving patients’ lives. Using MRI fat fraction to detect change early, together with strength testing to confirm it matters to patients, is the most responsible approach.

## Limitations and Future Work:

Describe the limitations of your project, including limitations of the sample size, available visits, MRI measurements, segmentation, biomarkers, functional measures, and analysis methods.

If you had more time or additional data, explain how you would improve the analysis. Examples include analyzing more participants, adding more timepoints, improving segmentation, comparing additional muscles, validating the MRI measure against functional outcomes, or testing whether the measure is sensitive to treatment response.

A major limitation of our study was having a small sample size. Although the clinical dataset included 34 participants with FSHD, only 21 had the complete MRI fat fraction and muscle strength measurements needed for our analysis. This limits how well our findings shows the larger FSHD population. Additionally, participants were followed for approximately two years with only three visits, which may not be enough to observe the full progression of a FSHD that often grows and progresses slowly over time.
Another limitation was that our clinical analysis focused only on the right tibialis anterior muscle, while our detailed MRI image analysis examined the left tibialis anterior muscle of a single patient, UW_0020, at two timepoints. Since FSHD can affect muscles differently and may progress unevenly between the two sides of the body, our results may not apply to other muscles or patients, and we can’t speak for other muscles that are affected by this disease. Our detailed MRI analysis also included only eight slices per visit and relied on the provided segmentation masks. Errors in identifying muscle boundaries or differences between slices could affect the calculated fat fraction values. Another limitation of our study is the use of linear interpolation to estimate fat fraction and muscle strength at common timepoints. Linear interpolation assumes that measurements change at a constant rate between visits. However, FSHD progression may not occur at a steady rate, since patients can experience periods of stability followed by faster changes. Therefore, our interpolated values may not fully represent the actual changes in muscle fat fraction and strength over time. Although we also used quadratic interpolation to compare the results, having more frequent measurements would help improve the accuracy of our estimates.Furthermore, our analysis focused only on MRI fat fraction and muscle strength, so it did not include other biomarkers or functional measurements that could provide additional information about disease progression. For future work, we would include more patients, so we can have more time to follow them over for a longer period with additional visits to better understand how FSHD progresses. We would also examine additional muscle groups to determine whether the same patterns occur throughout the body. In the future studies, we would also compare MRI fat fraction and muscle strength with other biomarkers and functional measurements to determine which methods best show the disease progression. Finally, studying patients before and after treatment could help researchers determine whether MRI fat fraction or muscle strength is more useful for detecting treatment-related changes in FSHD clinical trials.
